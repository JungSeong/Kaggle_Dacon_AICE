#### **[해설 및 핵심 이론] 문제 6**


* **상세 해설**
  
    **데이터 보존**: `copy()`를 사용하여 원본 데이터프레임을 복사해 둡니다.  
    **수치형 결측치 (자동화 및 벡터화)**: `for` 반복문을 사용하는 대신, `select_dtypes(include='number')`로 모든 수치형 컬럼을 한 번에 선택합니다. 그 후 `fillna()`에 평균값(`mean()`)을 전달하여, 코드 한 줄로 모든 수치형 결측치를 평균으로 대체합니다. 이는 코드를 간결하게 만들고 처리 속도를 높여줍니다.  
    **범주형 결측치**: `replace(None, '값')` 함수를 사용하여, 비어있는 값(None/NaN)을 지정된 문자열('Line-A', 'Type-Z')로 직관적으로 변경합니다.

```python
# 원본 보존을 위해 데이터프레임을 복사
yield_pre = yield_df.copy()

# 모든 수치형 컬럼 평균으로 채우기
yield_pre[num_cols] = yield_pre[num_cols].fillna(yield_pre[num_cols].mean())

# replace(찾을값, 바꿀값) 형식을 사용
yield_pre['production_line'] = yield_pre['production_line'].replace(np.nan, 'Line-A')
yield_pre['material_type'] = yield_pre['material_type'].replace(np.nan, 'Type-Z')

# 결과 확인
print(yield_pre.isnull().sum())
```

```python
# StandardScaler 객체 생성
scaler = StandardScaler()

# 훈련 데이터의 수치형 컬럼(num_scaling_cols)에 대해 스케일러를 학습(fit)하고 변환(transform)
X_train[num_scaling_cols] = scaler.fit_transform(X_train[num_scaling_cols])
# 검증 데이터의 수치형 컬럼(num_scaling_cols)에 대해 훈련 데이터에서 학습한 스케일러로만 변환(transform)
X_valid[num_scaling_cols] = scaler.transform(X_valid[num_scaling_cols])

# 수치형 컬럼만 스케일링된 전체 훈련 데이터를 X_train_scaled에 저장 (원-핫 인코딩된 컬럼은 그대로 유지됨)
X_train_scaled = X_train
# 수치형 컬럼만 스케일링된 전체 검증 데이터를 X_valid_scaled에 저장
X_valid_scaled = X_valid
```

    `Sequential` 모델을 사용하여 입력층, 2개의 은닉층, 출력층으로 구성된 딥러닝 모델을 설계합니다. 회귀 문제이므로 출력층의 노드는 1개이며, 손실 함수로는 `mean_squared_error`를 사용합니다. `fit` 함수로 모델을 학습시키며, `validation_data`를 지정하여 매 에포크마다 검증 성능을 함께 모니터링합니다.

```python
# EarlyStopping import
from tensorflow.keras.callbacks import EarlyStopping

# 모델 구조 정의
model = Sequential([
    Dense(64, activation='relu', input_shape=(X_train_scaled.shape[1],)),
    Dense(32, activation='relu'),
    Dense(16, activation='relu'),
    Dense(1)
])

# 모델 컴파일
model.compile(optimizer='adam', loss='mean_squared_error')

# Early Stopping 설정
early_stop = EarlyStopping(monitor='val_loss', 
                           patience=10, 
                           restore_best_weights=True)

# 모델 학습 (history 변수에 저장)
history = model.fit(
    X_train_scaled, y_train,
    epochs=100,
    batch_size=32,
    validation_data=(X_valid_scaled, y_valid),
    callbacks=[early_stop]
)
```
